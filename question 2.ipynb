```
%%sql -r dataframe_7
CREATE OR REPLACE TABLE SALES (
    SALE_ID INTEGER,
    PRODUCT VARCHAR(50),
    CUSTOMER VARCHAR(50),
    REGION VARCHAR(30),
    AMOUNT NUMBER(10,2)
);
```

```
%%sql -r dataframe_8
CREATE OR REPLACE STAGE SALES_STAGE;
```

```
%%sql -r dataframe_9
INSERT INTO SALES VALUES
(1, 'Laptop', 'Rahul', 'South', 55000),
(2, 'Mobile', 'Priya', 'North', 25000),
(3, 'Tablet', 'Arjun', 'East', 18000),
(4, 'Headphones', 'Sneha', 'West', 5000),
(5, 'Keyboard', 'Kiran', 'South', 2500);
```

```
%%sql -r dataframe_10
SELECT * FROM SALES;
```
