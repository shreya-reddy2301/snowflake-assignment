```
%%sql -r dataframe_6
SET recovery_time = CURRENT_TIMESTAMP();
```

```
%%sql -r dataframe_7
DELETE FROM RECOVERY_DEMO
WHERE ID IN (3, 4);
```

```
%%sql -r dataframe_8
SELECT * FROM RECOVERY_DEMO
ORDER BY ID;
```

```
%%sql -r dataframe_9
SELECT *
FROM RECOVERY_DEMO
AT (TIMESTAMP => $recovery_time)
WHERE ID IN (3, 4);
```
