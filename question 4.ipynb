```
%%sql -r dataframe_3
CREATE OR REPLACE TABLE TIME_TRAVEL_DEMO (
    ID INTEGER,
    NAME VARCHAR(50),
    CITY VARCHAR(50)
);
```

```
%%sql -r dataframe_4
INSERT INTO TIME_TRAVEL_DEMO VALUES
(1, 'Ravi', 'Chennai'),
(2, 'Anu', 'Bangalore'),
(3, 'Kiran', 'Hyderabad'),
(4, 'Sneha', 'Tirupati');
```

```
%%sql -r dataframe_5
SELECT * FROM TIME_TRAVEL_DEMO;
```

```
%%sql -r dataframe_6
SET before_changes = CURRENT_TIMESTAMP();
```

```
%%sql -r dataframe_7
UPDATE TIME_TRAVEL_DEMO
SET CITY = 'Mumbai'
WHERE ID = 1;
```

```
%%sql -r dataframe_8
DELETE FROM TIME_TRAVEL_DEMO
WHERE ID = 4;
```

```
%%sql -r dataframe_9
SELECT * FROM TIME_TRAVEL_DEMO;
```

```
%%sql -r dataframe_10
SELECT *
FROM TIME_TRAVEL_DEMO
AT (TIMESTAMP => $before_changes);
```

```
%%sql -r dataframe_11
SELECT COUNT(*) AS ORIGINAL_RECORD_COUNT
FROM TIME_TRAVEL_DEMO
AT (TIMESTAMP => $before_changes);
```
